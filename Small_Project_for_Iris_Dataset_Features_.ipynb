```python

books = []

for i in range(1,5):
  url = f"https://books.toscrape.com/catalogue/page-{i}.html"
  response = requests.get(url)
  response = response.content
  soup = BeautifulSoup(response, 'html.parser')
  ol = soup.find('ol')
  articles = ol.find_all('article', class_='product_pod')
  for article in articles:
    image = article.find('img')
    title = image.attrs['alt']
    starTag = article.find('p')
    star = starTag['class'][1]
    price = article.find('p', class_='price_color').text
    price = float(price[1:])
    books.append([title, star, price])

df = pd.DataFrame(books, columns=['Title', 'Star Rating', 'Price'])
df = pd.read_csv("Iris.csv")
df
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
# Load dataset from the csv file
dataset = pd.read_csv("Iris.csv")
# Create a figure and arss
fig, arss = plt.subplots(1, 4, figsize=(15, 5))

# Plot histograms
arss[0].hist(dataset.select_dtypes(include=np.number).columns, bins=15, color='blue', alpha=0.8)
arss[0].set_title('Histogram 1')

arss[1].hist(dataset.select_dtypes(include=np.number).columns, bins=15, color='green', alpha=0.8)
arss[1].set_title('Histogram 2')

arss[2].hist(dataset.select_dtypes(include=np.number).columns, bins=20, color='red', alpha=0.8)
arss[2].set_title('Histogram 3')

arss[3].hist(dataset.select_dtypes(include=np.number).columns, bins=20, color='purple', alpha=0.8)
arss[3].set_title('Histogram 4')
plt.xlabel('Values')
plt.ylabel('Frequency of the dataset')

# Adjust layout and show the plot
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
# Load dataset from the csv file
dataset = pd.read_csv("Iris.csv")
# Define colors
colors = ['Green', 'Blue', 'maroon', 'Gray']
# Plot the histogram
plt.figure(figsize=(8, 6))
for i, column in enumerate(dataset.columns[:-1]): plt.hist(dataset[column], bins=15, alpha=0.6, label=column, color=colors[i])

# Add labels and title
plt.xlabel('Value')
plt.ylabel('Frequency of the dataset')
plt.title('Histogram of Iris Dataset Features')
plt.legend()
plt.show()
```
